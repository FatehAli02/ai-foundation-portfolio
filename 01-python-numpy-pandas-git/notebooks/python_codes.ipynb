### E-Commerce System

```python
class ECommerece:
    def __init__(self):
        self.cart = []
    
    def Dispaly_store(self):
        headings = ["Item ID", "Name", "Price", "Stock"]
        print(" ".join(heading.ljust(15) for heading in headings))
        for row in store:
            print(" ".join(str(item).ljust(15) for item in row))

    def add_items(self):
        item_id =int(input("Enter ID of Item:"))
        found_item = None
        for item in store:
            if item[0] == item_id:
                found_item = item
                break
        if found_item:
            try:
                quantity = int(input("Enter quantity: "))

                if quantity <= 0:
                    print("Quantity must be greater than zero.")

                elif quantity > found_item[3]:
                    print("Not enough stock available.")

                else:   
                    total = found_item[2] * quantity
                    self.cart.append([found_item[1], quantity, found_item[2], total])

                    found_item[3] -= quantity

                    print(f"{found_item[1]} added to cart!")

            except ValueError:
                print("Please enter a valid quantity.")
           
        else:
            print("Item  not Present.")
    
    def calculate_bill(self):
        bill = 0
        for item in self.cart:
            bill += item[3]
        return bill
    
    def generate_Bill(self):
        headings = ["Name", "Quantity", "Price", "Total"]
        print(" ".join(heading.ljust(15) for heading in headings))
        for row in self.cart:
            print(" ".join(str(item).ljust(15) for item in row))
        print("Total Bill:",self.calculate_bill())




store = [
    [1, "Laptop", 800, 10],
    [2, "Smart Phone", 150, 15],
    [3, "Head Phone", 50, 20],
    [4, "Keyboard", 100, 25],
    [5, "Monitor", 80, 12]
]

cart = []
order = ECommerece()
while True:
    print("Welcome to E-Commerece store:")
    print("1. Store Contents")
    print("2. Add Items to cart")
    print("3. Calculate Bill")
    print("4. Genarate Bill")
    print("5. Exit E-Commerece")

    try:
        choice = int(input("Select Option: "))
    except ValueError:
        print("Please enter a valid menu number.")
        continue

    if (choice == 1):
        order.Dispaly_store()
    
    elif (choice == 2):
        order.add_items()
    
    elif (choice == 3):
        print("Total Bill:",order.calculate_bill())
    
    elif (choice == 4):
        order.generate_Bill()
    
    elif (choice == 5):
        print("Thank You !")
        break

    else:
        print("Inavlid Choice!")
```

### Student Management System

```python
class Student():
    def __init__(self,student_id, name, age, grade):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.grade = grade

    def show_student(self):
        print("-----------------------------------------------")
        print("Student Id: ",self.student_id)
        print("Name:       ",self.name)
        print("Age:        ",self.age)
        print("Grade:      ",self.grade)
        print("-----------------------------------------------")

class S_M_S(Student):
    def __init__(self,students):
        self.students = students
    def add(self):
        try:
            student_id = input("Enter the ID of Student: ")

            if student_id in self.students:
                print("Student with this ID already exists!")
                return

            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            grade = input("Enter Grade: ")

            if age <= 0:
                print("Age must be greater than zero.")
                return

            self.students[student_id] = [name, age, grade]
            print("New student record added!")

        except ValueError:
            print("Invalid age entered!")
        
    def read(self):
        headings = ["ID", "Name", "Age", "Grade"]
        print(" ".join(heading.ljust(15) for heading in headings))

        for key, value in self.students.items():
            print(" ".join(str(data).ljust(15) for data in [key] + value))
        
        return
    
    def update(self):
        student_id = input("Enter Id of Student To Update:")

        if student_id in self.students:
            while True:
                choice = int(input("Press 1 for name, 2 for Age, 3 for grade, 4 to exit update:") )
                if (choice == 1):
                    name2 = input("Enter New Name:")
                    self.students[student_id][0] = name2
                
                elif (choice == 2):
                    age2 = int(input("Enter New Age:"))
                    self.students[student_id][1] = age2

                elif (choice == 3):
                    grade2 = input("Enter New Grade:")
                    self.students[student_id][2] = grade2

                elif (choice == 4):
                    print("\nStudent Record Updated!\n")
                    break
                else:
                    print("Invalid Choice!")

    def delete(self):
        student_id = input("Enter Id to delete record:")
        if student_id in self.students:
            self.students.pop(student_id)
            print("\nStudent Record Deleted\n")
        else:
            print("Student not found!")

students = {
    "1" : ["Fateh Ali", 18, "A"],
    "2" : ["Asad", 17, "B"],
    "3" : ["Ahmad", 19, "C"],
}

sys = S_M_S(students)

while True:
    print("\n>>>> STUDENT Management System <<<<\n")
    print("1. Add Student Record")
    print("2. View Students ")
    print("3. Update Student Record")
    print("4. Delete Student Record")
    print("5. Display Specific Student")
    print("6. Exit System\n")

    try:
        option = int(input("Enter Number of Option: "))
    except ValueError:
        print("Please enter a valid number.")
        continue

    if (option == 1):
        sys.add()

    elif (option == 2):
        sys.read()

    elif (option == 3):
        sys.update()

    elif (option == 4):
        sys.delete()
    elif (option == 5):
        roll = input("Enter Id:")
        if roll in students:
            stu = Student(roll, students[roll][0],students[roll][1],students[roll][2])
            stu.show_student()
    elif (option == 6):
        print("\nThank You !")
        break
    else:
        print("Invalid Option Entered.")
```

### Number Guessing Python Game.py

```python
import random
def generate_number():
    nu = random.randint(1,10)
    print("\n->The Number is Ready<-\n")
    return nu
n = generate_number()
 
while True:
   
    try:
        guess = int(input("Guess the number between 1 and 10 (-1 to exit): "))
    except ValueError:
        print("Please enter a valid integer.")
        continue
    if (guess==-1):
        break
    if not 1 <= guess <= 10:
        print("Please enter a number between 1 and 10.")
        continue
    if (n==guess):
        print("Correct")
        n=generate_number()
   
    elif (n>guess):
            print("Too Low")
    elif(n<guess):
            print("Too high")
    else:
        print("Invalid Input.")
```

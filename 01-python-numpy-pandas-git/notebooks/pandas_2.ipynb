```python
# Imports and data
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 105, 106, 107],
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Usman", "Hina", None],
    "Age": [20, 21, np.nan, 22, 20, 20, 21, 23],
    "Marks": ["85", "92", "76", "88", "67", "67", "N/A", "90"],
    "City": ["Lahore", "Karachi", "Lahore", None, "Multan", "Multan", "lahore ", "Islamabad"],
    "JoinDate": ["2024-01-15", "2024-02-20", "2024-03-05", "2024-03-18",
                 "2024-04-02", "2024-04-02", "2024-05-11", "2024-06-01"]
})
```

#### Detecting missing values

#### Handling missing values
- Dropping

```python
df.dropna() # drops any row that contains a missing value
```

```python
df.dropna(how="all") # Drops row where all values are missing
```

```python
df.dropna(subset="Name") #check specific column
```

```python
df.dropna(thresh=5)  # keep rows with at least 5 non-missing values
```

```python
df.dropna(axis=1) # Drops column instead of rows
```

- Filling

```python
df["Age"] = df["Age"].fillna(df["Age"].median)
df
```

```python
df["City"] = df["City"].fillna("Unknown")
df
```

```python
df["Marks"] = df["Marks"].ffill()    # copy the previous value downward
df
```

| Strategy | Typical use |
|---|---|
| Mean | Numeric columns without extreme outliers |
| Median | Numeric columns with outliers or skew |
| Mode | Categorical columns (`df["City"].mode()[0]`) |
| Constant (`"Unknown"`, `0`) | When missing itself is meaningful |
| `ffill()` / `bfill()` | Ordered data such as time series |

```python
# imports and data
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali Khan", "Sara Ahmed", "Ahmed Raza", "Fatima Noor", "Usman Ali", "Hina Malik"],
    "Age": [20, 21, 19, 22, 20, 21],
    "Marks": [85, 92, 76, 88, 67, 92],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Multan", "Lahore"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Email": ["ali.khan@uni.edu.pk", "sara.ahmed@uni.edu.pk", "ahmed.raza@gmail.com",
              "fatima.noor@uni.edu.pk", "usman.ali@gmail.com", "hina.malik@uni.edu.pk"]
})
df
```

#### Creating new columns

```python
df["Age_Next_Year"] = df["Age"] + 1
df
```

```python
df["Marks+Percentage"] = df["Marks"] / 100 * 100
df
```

- Conditional Columns with np.where()

```python
df["Result"] = np.where(df["Marks"] >= 80, "High", "Standard")
df
```

- Multiple conditions with np.select()

```python
conditions = [df["Marks"] >= 90, df["Marks"] >= 80, df["Marks"] >= 70]
choices = ["A", "B", "C"]
df["Grade"] = np.select(conditions, choices, default="D")
df
```

- Binding numbers with pd.cut()

```python
df["Band"] = pd.cut(df["Marks"], bins=[0,70,85,100], labels=["Low", "Mid", "High"])
df
```

- Other ways to add columns

```python
df = df.assign(Double_Marks=df["Marks"]*2)
df
```

```python
df.insert(1, "Country", 'Pakistan')
df
```

#### map()

```python
# map() works on a Series only and transforms each value using a dictionary or a function.
df["Gender_Full"] = df["Gender"].map({"M" : 'Male', "F" : "Female"})
df
```

```python
df["Name_Length"] = df["Name"].map(len)
df
```

#### replace()

```python
# replace() swaps specific values and leaves everything else unchanged. That is the key difference from map()
df["City"].replace("Islamabad", "ISB")
```

```python
df["City"].replace({"Lahore" : "LHR", "Karachi" : "KHI"})
```

```python
df.replace({"City" : {"Multan": "MUX"}})
```

```python
# Remember to assign the result back, because replace() returns a new object.
```

#### apply()

```python
# On a Series
def pass_fail(m):
    return "Pass" if m >= 70 else 'Fail'

df["Status"] = df["Marks"].apply(pass_fail)
df
```

```python
df["Marks_Squared"] = df["Marks"].apply(lambda x: x ** 2)
df
```

```python
# On DataFrame
df["Summary"] = df.apply(lambda row: f"{row["Name"]} from {row["City"]}", axis=1)
df
# axis = 0 -> the ftn recieves each column 
# axis = 1 -> it recieves each row
```

### 6. Choosing the right tool

| Task | Best tool |
|---|---|
| Arithmetic between columns | Direct operators (`df["A"] + df["B"]`) |
| Simple if/else | `np.where()` |
| Several conditions | `np.select()` |
| Numeric ranges into labels | `pd.cut()` |
| Recode every value with a lookup | `map()` |
| Fix a few specific values | `replace()` |
| Text cleanup and extraction | `.str` methods |
| Custom logic that nothing else handles | `apply()` |

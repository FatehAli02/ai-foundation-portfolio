```python
# Imports and data
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 105, 106, 107],
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Usman", "Hina", None],
    "Age": [20, 21, np.nan, 22, 20, 20, 21, 23],
    "Marks": ["85", "92", "76", "88", "67", "67", "N/A", "90"],
    "City": ["Lahore", "Karachi", "Lahore", None, "Multan", "Multan", "lahore ", "Islamabad"],
    "JoinDate": ["2024-01-15", "2024-02-20", "2024-03-05", "2024-03-18",
                 "2024-04-02", "2024-04-02", "2024-05-11", "2024-06-01"]
})
```

```python
df.info()
```

#### Detecting missing values

```python
df.isna() # Returns true or false for each cell
```

```python
df[df['Name'].isna()]
```

```python
df.notna() # Opposite of isna()
```

```python
df.isna().sum() # Missing count per column
```

#### Handling missing values
- Dropping

```python
df.dropna() # drops any row that contains a missing value
```

```python
df.dropna(how="all") # Drops row where all values are missing
```

```python
df.dropna(subset="Name") #check specific column
```

```python
df.dropna(thresh=5)  # keep rows with at least 5 non-missing values
```

```python
df.dropna(axis=1) # Drops column instead of rows
```

- Filling

```python
df["Age"] = df["Age"].fillna(df["Age"].median)
df
```

```python
df["City"] = df["City"].fillna("Unknown")
df
```

```python
df["Marks"] = df["Marks"].ffill()    # copy the previous value downward
df
```

| Strategy | Typical use |
|---|---|
| Mean | Numeric columns without extreme outliers |
| Median | Numeric columns with outliers or skew |
| Mode | Categorical columns (`df["City"].mode()[0]`) |
| Constant (`"Unknown"`, `0`) | When missing itself is meaningful |
| `ffill()` / `bfill()` | Ordered data such as time series |

#### Duplicates

```python
df.duplicated() # Returns True for duplicate row after the first
```

```python
df.duplicated().sum() # Counts the duplicate rows
```

```python
df[df.duplicated(keep=False)] # Returns the Duplicate rows, includeing first
```

```python
df.drop_duplicates() # removes the duplicate row
```

```python
df.drop_duplicates(subset=["ID"]) # duplicates checked by ID only
```

```python
df.drop_duplicates(subset=["ID"], keep="last") # keep the last copy instead
```

#### Data types

```python
# astype() : simple conversion
df["ID"] = df["ID"].astype("int32")
```

```python
df["City"] = df["City"].astype("category")
```

```python
# pd.to_numeric(): safe numeric conversion

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# With errors="coerce", anything that can’t be parsed becomes NaN
df
```

```python
# pd.to_datetime(): dates

df["JoinDate"] = pd.to_datetime(df["JoinDate"])

df["JoinDate"].dt.year
```

#### Cleaning text

```python
df["City"] = df["City"].str.strip()     # remove leading/trailing spaces
df["City"] = df["City"].str.title()     # "lahore" -> "Lahore"
df["City"].unique()                     # verify the distinct values
```

```python
df["City"].value_counts()               # count each value
```

#### Renaming

```python
df = df.rename(columns={"JoinDate": "Join_Date"})
df
```

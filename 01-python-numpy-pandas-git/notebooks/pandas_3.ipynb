```python
# Imports and data

import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Hina", "Bilal", "Ayesha", "Zain", "Maryam"],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Multan",
             "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F", "M", "F"],
    "Department": ["CS", "CS", "EE", "CS", "EE", "BBA", "BBA", "EE", "CS", "BBA"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 19, 22],
    "Marks": [85, 92, 76, 88, 67, 92, 71, 80, 95, 84]
})
df
```

#### groupby()
- Three Steps:
    - Split the rows into groups based on a column.
    - Apply a calculation (mean, sum, count) to each group.
    - Combine the results into a new object.

```python
df.groupby("City")["Marks"].mean()
```

```python
# Other common aggreagte methods can also be used

# .sum() , .max(), .min(), .count(), .median()

df.groupby("City").size()

# count() ignores missing values, while size() counts every row.
```

- Grouping by multiple columns

```python
df.groupby(["City", "Gender"])["Marks"].mean()
```

- Selecting several columns

```python
df.groupby("City")[["Age", 'Marks']].mean()
```

- Group keys as the index

```python
df.groupby("City", as_index=False)["Marks"].mean()
```

#### agg(): several calculations at once

```python
df.groupby('City')['Marks'].agg(['mean', 'min', 'max', 'count'])
```

- Diff functions per column

```python
df.groupby('City').agg({
    "Marks" : ["mean", "max"],
    "Age" : "mean"
})
```

- Named aggregations

```python
df.groupby("City").agg(
    Avg_Marks=('Marks', "mean"),
    Top_Marks=("Marks", "max"),
    Students=("Name", "count")
)
```

```python
df.groupby("City").agg(Mark_Rnage=("Marks", lambda s: s.max() - s.min()))
```

#### transform() and filter()
- transform() returns a result with the same length as the original DataFrame
- filter() keeps or drops whole groups based on a condition

```python
df["City_Avg"] = df.groupby("City")["Marks"].transform("mean")
df
```

```python
df.groupby("City").filter(lambda g: len(g) >= 3)  # only cities with 3+ students
```

#### value_counts()
- Counts how often each value appears in a column

```python
df["City"].value_counts()
```

```python
df["City"].value_counts(normalize=True)             # proportions instead of counts
```

```python
df["City"].value_counts(dropna=False)               # include missing values
```

```python
df[["City", "Gender"]].value_counts()
```

#### Sorting grouped results

```python
df.groupby('City')["Marks"].mean().sort_values(ascending=False)
```

```python
df.groupby("City")["Marks"].mean().idxmax() # which city has highest average
```

#### pd.crosstab()
- Counts how often combinations of two columns occur, laid out as a table

```python
pd.crosstab(df["City"], df["Gender"])
```

```python
pd.crosstab(df["City"], df["Gender"], margins=True)          # adds row and column totals
pd.crosstab(df["City"], df["Gender"], normalize="index")     # row proportions
```

#### Pivot Tables\
- A pivot table reshapes grouped data into a spreadsheet-style summary, with one grouping down the rows and another across the columns.

```python
df.pivot_table(values="Marks", index="City", columns="Gender", aggfunc="mean")
```

```python
df.pivot_table(values=["Marks", "Age"], index="Department", aggfunc="mean")
```

```python
df.pivot_table(values="Marks", index="Department", aggfunc=["mean", "max"])
```

```python
# imports and data
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "StudentID": [1, 2, 3, 4, 5],
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman"],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Multan"]
})

marks = pd.DataFrame({
    "StudentID": [1, 2, 3, 6],
    "Subject": ["Math", "Physics", "Math", "English"],
    "Score": [85, 92, 76, 88]
})

fees = pd.DataFrame({
    "ID": [1, 2, 4, 5],
    "Fee_Paid": [True, True, False, True]
})
```

#### merge(): combining tables by key

```python
# Like inner join in SQL 
pd.merge(students, marks, on='StudentID')
```

```python
# left join / right join
pd.merge(students, marks, on="StudentID", how="left")
```

```python
# outer join
pd.merge(students, marks, on="StudentID", how="outer")
```

```python
# If diff key / column names
pd.merge(students, fees, left_on="StudentID", right_on="ID")
```

```python
# Dropping redundant one
pd.merge(students, fees, left_on="StudentID", right_on="ID").drop(columns="ID")
```

Overlapping column names

If both tables have a non-key column with the same name, Pandas adds suffixes _x and _y.

```python
# pd.merge(a, b, on="StudentID", suffixes=("_mid", "_final"))
```

```python
# Merging on index
students.set_index("StudentID").join(marks.set_index("StudentID"), how="left")
```

```python
# Debugging of merges
pd.merge(students, marks, on="StudentID", how="outer", indicator=True)
```

#### concat(): stacking tables

```python
batch1 = pd.DataFrame({"Name" : ["Ali", "Sara"], "Marks" : [85,92]})
batch2 = pd.DataFrame({"Name" : ["Ahmad", "Fatima"], "Marks" : [76, 88]})

pd.concat([batch1, batch2]) # Vertical
```

```python
pd.concat([batch1, batch2], ignore_index=True)
```

```python
pd.concat([batch1, batch2], axis=1)
```

```python
# If columns don't match, missing will be filled with Nan

batch3 = pd.DataFrame({"Name" : ["Usman"], "City" : ["Multan"]})
pd.concat([batch1, batch3], ignore_index=True)
```

#### Date and Time operations

```python
dates = pd.DataFrame({
    "Student": ["Ali", "Sara", "Ahmed", "Fatima", "Usman"],
    "JoinDate": ["2024-01-15", "2024-02-20", "2024-03-05", "2024-03-18", "2024-04-02"],
    "Marks": [85, 92, 76, 88, 67]
})
dates["JoinDate"] = pd.to_datetime(dates["JoinDate"])
dates["JoinDate"]
```

```python
pd.to_datetime("05/03/2024", dayfirst=True)  # 5 March 2024
```

```python
pd.to_datetime("05-03-2024", format="%d-%m-%Y") # explicit format, safest
```

- .dt

```python
dates["Year"] = dates['JoinDate'].dt.year
dates["Day_Name"] = dates['JoinDate'].dt.day_name()
dates["Quarter"] = dates['JoinDate'].dt.quarter
dates["Is_Weekend"] = dates["JoinDate"].dt.dayofweek >= 5
```

```python
dates
```

- Date Arithmetics

```python
today = pd.Timestamp("2025-06-01")
dates["Days_Enrolled"] = (today - dates["JoinDate"]).dt.days
dates
```

- Filtering by dates

```python
dates[dates["JoinDate"].between("2024-02-01", "2024-03-31")]
```

```python
ts= dates.set_index("JoinDate").sort_index()
ts.loc["2024-03"]
```

```python
# Formatting date as text
dates["JoinDate"].dt.strftime("%d %b %Y")
```

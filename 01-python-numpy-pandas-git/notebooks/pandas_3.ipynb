```python
# Imports and data

import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Hina", "Bilal", "Ayesha", "Zain", "Maryam"],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Multan",
             "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F", "M", "F"],
    "Department": ["CS", "CS", "EE", "CS", "EE", "BBA", "BBA", "EE", "CS", "BBA"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 19, 22],
    "Marks": [85, 92, 76, 88, 67, 92, 71, 80, 95, 84]
})
df
```

#### groupby()
- Three Steps:
    - Split the rows into groups based on a column.
    - Apply a calculation (mean, sum, count) to each group.
    - Combine the results into a new object.

```python
df.groupby("City")["Marks"].mean()
```

```python
# Other common aggreagte methods can also be used

# .sum() , .max(), .min(), .count(), .median()

df.groupby("City").size()

# count() ignores missing values, while size() counts every row.
```

- Grouping by multiple columns

```python
df.groupby(["City", "Gender"])["Marks"].mean()
```

- Selecting several columns

```python
df.groupby("City")[["Age", 'Marks']].mean()
```

- Group keys as the index

```python
df.groupby("City", as_index=False)["Marks"].mean()
```

#### agg(): several calculations at once

```python
df.groupby('City')['Marks'].agg(['mean', 'min', 'max', 'count'])
```

- Diff functions per column

```python
df.groupby('City').agg({
    "Marks" : ["mean", "max"],
    "Age" : "mean"
})
```

- Named aggregations

```python
df.groupby("City").agg(
    Avg_Marks=('Marks', "mean"),
    Top_Marks=("Marks", "max"),
    Students=("Name", "count")
)
```

```python
df.groupby("City").agg(Mark_Rnage=("Marks", lambda s: s.max() - s.min()))
```

#### transform() and filter()
- transform() returns a result with the same length as the original DataFrame
- filter() keeps or drops whole groups based on a condition

```python
df["City_Avg"] = df.groupby("City")["Marks"].transform("mean")
df
```

```python
df.groupby("City").filter(lambda g: len(g) >= 3)  # only cities with 3+ students
```

#### value_counts()
- Counts how often each value appears in a column

```python
df["City"].value_counts()
```

```python
df["City"].value_counts(normalize=True)             # proportions instead of counts
```

```python
df["City"].value_counts(dropna=False)               # include missing values
```

```python
df[["City", "Gender"]].value_counts()
```

#### Sorting grouped results

```python
df.groupby('City')["Marks"].mean().sort_values(ascending=False)
```

```python
df.groupby("City")["Marks"].mean().idxmax() # which city has highest average
```

#### pd.crosstab()
- Counts how often combinations of two columns occur, laid out as a table

```python
pd.crosstab(df["City"], df["Gender"])
```

```python
pd.crosstab(df["City"], df["Gender"], margins=True)          # adds row and column totals
pd.crosstab(df["City"], df["Gender"], normalize="index")     # row proportions
```

#### Pivot Tables\
- A pivot table reshapes grouped data into a spreadsheet-style summary, with one grouping down the rows and another across the columns.

```python
df.pivot_table(values="Marks", index="City", columns="Gender", aggfunc="mean")
```

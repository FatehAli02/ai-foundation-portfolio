### Series

- A Series is a one-dimensional labeled array. Think of it as a single column of data, where every value has an associated index.

```python
import pandas as pd
marks = pd.Series([85, 79, 90, 54])
print(marks)
```

```python
# custom indexes

marks = pd.Series(
    [86, 90, 70],
    index= ['Physics', 'Math', 'Chemistry'],
    name= 'Student Marks'
)
print(marks)
print()
print(marks['Math'])
```

```python
# Series from list
country = ['India', 'Pakistan', 'USA', 'Nepal', 'Turkey']
pd.Series(country)
```

```python
# Series from dict
marks = {
    'maths' : 67,
    'english' : 57,
    'science' : 100,
    'urdu' : 77
}
marks_series = pd.Series(marks, name="Fateh's Result")
marks_series
```

### DataFrame

- A DataFrame is a two-dimensional labeled data structure consisting of rows and columns, similar to an Excel table or SQL table.

```python
# It can be created by dict and list a well
data = {
    "Name" : ["Fateh", "Ali", "Asad", "Zain"],
    "Age" : [20, 30, 21, 24],
    "Marks": [85, 92, 76, 88],
    "City" : ["Lahore", "Karachi", "Islamabd", "Multan"]
}

df = pd.DataFrame(data)
df
```

### Selecting Columns

```python
# single column
df['Name'] # Gives a Series
```

```python
# multiple column
df[['Name', 'Marks']] # give a DataFrame
```

### Row Indexing

```python
# loc selects data using row and column labels.
# With loc, label-based slices include the ending label when present.
df.loc[1]
```

```python
df.loc[1,'Name']
```

```python
df.loc[0:2, 'Name']
```

```python
df.loc[:, ['Name', 'Age']]
```

```python
# iloc selects data using integer positions.
# With iloc, the ending position is excluded, just like ordinary Python slicing.

df.iloc[1]
```

```python
df.iloc[1,0]
```

```python
df.iloc[0:2]
```

```python
df.iloc[:, 0:2]
```

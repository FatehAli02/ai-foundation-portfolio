#### **Understanding and implementing a regression model**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=15,
    bias=50,
    random_state=42
)

print("Shape of X: ", X.shape)
print("Shape of y: ", y.shape)
```

```python
# Data Visualization

plt.scatter(X, y)

plt.xlabel("X (inputs)")
plt.ylabel("y (targets)")
plt.title("Dataset")

plt.show()
```

```python
# Training model
model = LinearRegression()

model.fit(X, y)
```

```python
w = model.coef_
b = model.intercept_

print("Weight w: ", w)
print("Bias b: ", b)
```

```python
X_line = np.linspace(X.min(), X.max(),100).reshape(-1,1)

prediction_line = model.predict(X_line)

plt.scatter(X, y, label="Actual Data", color="green")
plt.plot(X_line, prediction_line, color="black", label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")

plt.legend()
plt.show()
```

```python
test_value = np.array([[5]])

prediction = model.predict(test_value)

print("Prediction: ", prediction)
```

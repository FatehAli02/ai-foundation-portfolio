#### **Gradient Descent**
*Implementation*

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1,2,3,4])
y = np.array([150, 250, 350, 450])
```

```python
w = 0 # weight
b = 0 # bias
a = 0.1 # learning rate
iterations = 1000
```

```python
# Cost Function
def cost_function(X, y, w, b):

    m = len(y)

    predictions = w * X + b

    errors = predictions - y
    squared_errors = errors ** 2
    
    cost = np.sum(squared_errors) / (2 * m)

    return cost
```

```python
# Calculating Gradient
def gradient(X, y, w, b):
    m = len(y)

    predictions = w * X + b

    errors = predictions - y
    dj_dw = np.sum(errors * X) / m
    dj_db = np.sum(errors) / m

    return dj_dw, dj_db
```

```python
# Looping and updating gradient descent
def gradient_descent(X, y, w, b, a, iterations):

    costs = []
    for i in range(iterations):

        dj_dw, dj_db = gradient(X, y, w, b)

        w = w - a * dj_dw
        b = b - a * dj_db
        cost = cost_function(X, y, w, b)
        costs.append(cost)
    
    print("Latest Weight: ", w)
    print("Latest Bias: ", b)
    print("Latest Cost: ", costs[-1])

    return costs, w, b
```

```python
costs, w_new, b_new = gradient_descent(X, y, w, b, a, iterations)
```

```python
plt.plot(costs)

plt.xlabel("Iterations")
plt.ylabel("Cost")

plt.show()
```

```python
plt.scatter(X, y, label="Actual Data")

predictions = w_new * X + b_new

plt.plot(X, predictions, label="Gradient Descent Line")

plt.xlabel("X")
plt.ylabel("y")

plt.legend()
plt.show()
```

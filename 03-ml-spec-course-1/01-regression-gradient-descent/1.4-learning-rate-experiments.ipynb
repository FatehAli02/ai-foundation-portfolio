#### **Learning Rate Experiments**

Understanding how the learning rate (a) affects Gradient Descent.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1,2,3,4])
y = np.array([150, 250, 350, 450])
```

```python
# Cost Function
def cost_function(X, y, w, b):

    m = len(y)

    predictions = w * X + b

    errors = predictions - y
    squared_errors = errors ** 2
    
    cost = np.sum(squared_errors) / (2 * m)

    return cost
```

```python
# Calculating Gradient
def gradient(X, y, w, b):
    m = len(y)

    predictions = w * X + b

    errors = predictions - y
    dj_dw = np.sum(errors * X) / m
    dj_db = np.sum(errors) / m

    return dj_dw, dj_db
```

```python
# Training Function
def gradient_descent(X, y, a, iterations):

    costs = []
    w = 0
    b = 0
    for i in range(iterations):

        dj_dw, dj_db = gradient(X, y, w, b)

        w = w - a * dj_dw
        b = b - a * dj_db
        cost = cost_function(X, y, w, b)
        costs.append(cost)

    return costs, w, b
```

```python
iterations = 300

cost_1, w_1, b_1 = gradient_descent(X, y, 0.001, iterations)

cost_2, w_2, b_2 = gradient_descent(X, y, 0.01, iterations)

cost_3, w_3, b_3 = gradient_descent(X, y, 0.1, iterations)
```

```python
print("Results")

print("Learning Rate: 0.001")
print("Weight 1: ", w_1)
print("Bias 1: ", b_1)
print("Cost 1:", cost_1[-1])
print()

print("Learning Rate: 0.01")
print("Weight 1: ", w_2)
print("Bias 1: ", b_2)
print("Cost 1:", cost_2[-1])
print()

print("Learning Rate: 0.1")
print("Weight 1: ", w_3)
print("Bias 1: ", b_3)
print("Cost 1:", cost_3[-1])
```

```python
plt.plot(cost_1, label="Alpha = 0.001")
plt.plot(cost_2, label="Alpha = 0.01")
plt.plot(cost_3, label="Alpha = 0.1")

plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Learning Rate Comparison")

plt.legend()
plt.show()
```

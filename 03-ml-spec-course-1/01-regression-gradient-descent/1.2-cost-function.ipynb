#### **Cost Function**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
inputs = np.array([1, 2, 3, 4])
actual = np.array([150, 250, 350, 450])
```

```python
# Cost Function

def cost_function(X, y, w, b):

    m = len(y)

    predictions = w * X + b

    errors = predictions - y
    squared_errors = errors ** 2
    
    cost = np.sum(squared_errors) / (2 * m)

    return cost
```

```python
print("Cost 1:", cost_function(inputs, actual, 100, 50))
print("Cost 2:", cost_function(inputs, actual, 80, 30))
print("Cost 3:", cost_function(inputs, actual, 50, 20))
```

```python
weights = np.arange(0,201, 50)

costs = []

for w in weights:
    cost = cost_function(inputs, actual, w, 50)
    costs.append(cost)

plt.plot(weights, costs)

plt.xlabel("Weight (w)")
plt.ylabel("Cost J(w,b)")
plt.title("Cost Function vs Weight")

plt.show()
```

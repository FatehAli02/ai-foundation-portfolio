#### **Feature Scaling**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset with 2 features 

X = np.array([
    [1000, 2], [1500, 3], [2000, 3], [3000, 4], [4000, 5]
])

print(X)
```

```python
print("Minimum: ", np.min(X, axis=0))
print("Maximum: ", np.max(X, axis=0))
print("Mean: ", np.mean(X, axis=0))
print("Standard Deviation: ", np.std(X, axis=0))
```

**Mean Normalization - Technique 1**

```python
def mean_normalization(X):
    mean = np.mean(X, axis=0)
    max = np.max(X, axis=0)
    min = np.min(X, axis=0)

    x_mn = (X - mean) / (max - min)

    return x_mn
```

**Z-score Normalization - Technique 2**


```python
def zscore_normalization(X):
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)

    x_zn = (X - mean) / std

    return x_zn
```

**Min Max Scaling - Technique 3**


```python
def min_max_scaling(X):
    max = np.max(X, axis=0)
    min = np.min(X, axis=0)

    x_mm = (X - min) / (max - min)

    return x_mm
```

 **Comparison**

```python
X_mean = mean_normalization(X)
X_zscore = zscore_normalization(X)
X_minmax = min_max_scaling(X)

print("Original Data:\n", X)

print("\nMean Normalization:\n", X_mean)

print("\nZ-score Normalization:\n", X_zscore)

print("\nMin-Max Scaling:\n", X_minmax)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(15, 4))

datasets = [
    (X, "Original Data"),
    (X_mean, "Mean Normalization"),
    (X_zscore, "Z-score Normalization"),
    (X_minmax, "Min Max Scaling")
]

for ax, (data, title) in zip(axes, datasets):
    ax.plot(data[:, 0], label="Feature 1")
    ax.plot(data[:, 1], label="Feature 2")
    ax.set_title(title)
    ax.legend()

plt.tight_layout()
plt.show()
```

```python
# imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**2D Line plot**
- Bivarite Analysis
- Should be on 2 columns
- Both Columns may be numerical and maybe 1 numerical and 1 categorical
- Use case: Time Series Data


```python
# Plotting Simple Function
price = [48000, 54000, 57000, 49000, 47000, 45000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)
```

```python
# Importing dataset
batsman = pd.read_csv('data/sharma-kohli.csv')
batsman
```

```python
# On X axis year, On Y axis Runs
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
#Plotting multiple lines
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# Labels and Titles
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
```

```python
# Colors(hex can also be used), style and line width
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=3) # e.g. 'dashed', 'dashdot'
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dotted', linewidth=2)

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
```

```python
# Point markers
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=3, marker='D', markersize=5)
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', linewidth=2, marker='>')

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
```

```python
# Naming lines
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=3, marker='D', markersize=5, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', linewidth=2, marker='>', label='Sharma')

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")

plt.legend(loc='best') # loc can be changed, best is by default
```

```python
# Limiting axes
# Outlier effected graph
price = [48000, 54000, 57000, 49000, 47000, 45000, 4500000]
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]

plt.plot(year, price)
```

```python
# Defining Range to avoid above condition
plt.plot(year, price)
plt.ylim(0, 100000)
plt.xlim(2017, 2019)
```

```python
# grid
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=3, marker='D', markersize=5, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', linewidth=2, marker='>', label='Sharma')

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")

plt.grid()
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=3, marker='D', markersize=5, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', linewidth=2, marker='>', label='Sharma')

plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel("Year")
plt.ylabel("Runs Scored")

plt.grid()
plt.show() #For plotting Graphs other than jupyter nbs
```

**Scatter Plots**

- Bivariate Analysis
- numerical vs numerical
- Use case: Finding correlation between 2 quantities

```python
x = np.linspace(-10, 10, 50)

y = 10*x + 3 + np.random.randint(0, 300, 50)
y
```

```python
plt.scatter(x, y)
```

```python
# scatter on dataset
batter = pd.read_csv('data/batter.csv')
batter = batter.head(50)
batter
```

```python
plt.scatter(batter['avg'], batter['strike_rate'], color='red', marker='+')
plt.title("Avg and Strike rate analysis of top 50 batters")
plt.xlabel("Average")
plt.ylabel("SR")
```

```python
# size
tips = sns.load_dataset('tips')
tips.head()

plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']*20)
```

```python
# scatterplot using plt.plot
# Faster Technique but doesn't provide some funtions
plt.plot(tips['total_bill'], tips['tip'], 'o')
```

**Bar Chart**

- Bivariate Analysis, Univariate
- Numerical vs Categorical
- Use Case: Aggregate analysis of groups

```python
childern = [10, 20 ,40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, childern, color='black')
```

```python
# Horizontal Bar Chart
plt.barh(colors, childern, color='black')
```

Colors and Labels can be controlled as 2d & scatter plots

```python
df = pd.read_csv('data/batsman_season_record.csv')
df
```

```python
# Multiple Bar charts
plt.bar(df['batsman'], df['2015'])
plt.bar(df['batsman'], df['2016'])
plt.bar(df['batsman'], df['2017'])
```

```python
# Side by side to avoid messy bar chart like above
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2, color='yellow')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, color='green')
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2, color='black')

# xticks
plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

```python
# Stacked Bar chart
plt.bar(df['batsman'], df['2017'], label='2017')
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], label='2016')
plt.bar(df['batsman'], df['2015'], bottom=df['2016'] + df['2017'], label='2015')

plt.legend()
plt.show()
```

**Pie Chart**

- Univariate / Bivariate
- Categorical vs Numerical
- Use Case: To find contribution on a standard scale

```python
data = [23, 45, 100, 20, 49]
subjects = ['eng', 'maths', 'urdu', 'sci', 'history']
plt.pie(data, labels=subjects)
plt.show()
```

```python
# With dataset
df = pd.read_csv('data/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', colors=['blue', 'pink', 'red', 'green', 'black', 'yellow'])
plt.show()
```

```python
#explode & shadow
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode= [0.3,0,0,0,0,0.1], shadow=True)
```

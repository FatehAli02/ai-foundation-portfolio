### **Advance Matplotlib**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Colored Scatterplots**

Color Encoding on 2D plot

```python
iris = pd.read_csv('data/iris.csv')
iris.head()
```

```python
#replcing target column values to numbers
iris['Species'] = iris['Species'].replace({'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2})
iris.sample(5)
```

```python
#cmap and alpha 
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

**Plot Size**

```python
plt.figure(figsize=(15,7)) # (x, y) -> x = width, y = height
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

**Annotations**

Placing text on graph

```python
batters = pd.read_csv('data/batter.csv')
batters.head()
```

```python
sample_batters = batters.head(100).sample(25, random_state=5)
sample_batters
```

```python
plt.figure(figsize=(20, 10))
plt.scatter(sample_batters['avg'], sample_batters['strike_rate'], s=sample_batters['runs'])
for i in range(sample_batters.shape[0]):
    plt.text(sample_batters['avg'].values[i], sample_batters['strike_rate'].values[i], sample_batters['batter'].values[i])
```

**Horizontal and Vertical Lines**

```python
plt.figure(figsize=(18, 10))
plt.scatter(sample_batters['avg'], sample_batters['strike_rate'], s=sample_batters['runs'])

plt.axhline(130, color='red')
plt.axhline(140, color='green')
plt.axvline(30, color='red')

for i in range(sample_batters.shape[0]):
    plt.text(sample_batters['avg'].values[i], sample_batters['strike_rate'].values[i], sample_batters['batter'].values[i])
```

**Subplots**

```python
# Diff way to plot graph
fig, ax = plt.subplots(figsize=(15, 6))
ax.scatter(batters['avg'], batters['strike_rate'], color='green')
ax.set_title('Comparison')
ax.set_xlabel('Average')
ax.set_ylabel('SR')
fig.show()
```

```python
# Two or more Scatter plots side by side
fig, axis = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10,6))
axis[0].scatter(batters['avg'], batters['strike_rate'], color='black')
axis[1].scatter(batters['avg'], batters['runs'], color='green')

axis[0].set_title('Avg vs Strike Rate')
axis[0].set_ylabel('SR')

axis[1].set_title('Avg vs Run')
axis[1].set_ylabel('Run')
axis[1].set_xlabel('Avg')
```

```python
# 4 plots
fig, axis = plt.subplots(nrows=2, ncols=2, figsize=(10,10))
axis[0,0].scatter(batters['avg'], batters['strike_rate'], color='black')
axis[0,1].scatter(batters['avg'], batters['runs'], color='green')

axis[1,0].hist(batters['avg'], color='black')
axis[1,1].hist(batters['runs'], color='green')
```

**3D Scatter Plots**

```python
fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(batters['runs'], batters['avg'], batters['strike_rate'])
ax.set_title('Batter Analysis')

ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

**3D Surface Plots**

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
```

```python
xx, yy = np.meshgrid(x, y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx, yy, z, cmap='winter')
fig.colorbar(graph)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx, yy, z, cmap='jet')
fig.colorbar(graph)
```

**Contour Plot**

3d graph to 2d

```python
# Actual Graph
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx, yy, z, cmap='winter')
fig.colorbar(graph)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()
graph = ax.contour(xx, yy, z, cmap='viridis')
fig.colorbar(graph)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()
graph = ax.contourf(xx, yy, z, cmap='viridis')
fig.colorbar(graph)
```

**Heatmap**

```python
delivery = pd.read_csv('data/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run'] == 6)]
df
```

```python
grid = df.pivot_table(index='overs', columns='ballnumber', values='batsman_run', aggfunc='count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.colorbar()
```

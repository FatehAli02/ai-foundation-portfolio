1. **Relational Plot**
- Bivariate Analysis
- Statistical Relation between 2 or more variables

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axis level function
sns.scatterplot(tips, x='total_bill', y='tip')
```

```python
# figure level func -> square shape
sns.relplot(tips, x='total_bill', y='tip', kind='scatter')
```

```python
# hue, style, size
sns.scatterplot(tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

**Line Plot**

```python
gap = px.data.gapminder()
df = gap[gap['country'] == 'Pakistan']
df
```

```python
# Axes level func
sns.lineplot(df, x= 'year', y='lifeExp')
```

```python
sns.relplot(df, x='year', y='lifeExp', kind='line')
```

```python
df = gap[gap['country'].isin(['Pakistan', 'Brazil', 'Germany'])]
df
```

```python
sns.relplot(kind='line', data=df, x='year', y='lifeExp', hue='country', style='continent')
```

```python
# facet plot -> figure level func -> only works with relplot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='smoker')
```

```python
sns.relplot(gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=4)
```

```python
# Figure level -> displot
# Axes level -> histplot, kdeplot, rugplot
```

```python
sns.histplot(tips, x='total_bill')
```

```python
sns.displot(tips,x='total_bill', kind='hist')
```

```python
sns.displot(tips,x='total_bill', kind='hist', bins=20)
```

```python
# It can be plotted on categorical data
sns.displot(tips,x='day', kind='hist')
```

```python
# hue parameter
sns.displot(tips,x='tip', kind='hist', hue='sex')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(titanic,x='age', kind='hist', element='step', hue='sex')
```

**Matrix Plot**

**Heatmap**

```python
df = gap.pivot(index='country', columns='year', values='lifeExp')
# axes level func
plt.figure(figsize=(15,15))
sns.heatmap(df)
```

```python
df = gap[gap['continent'] == 'Europe'].pivot(index='country', columns='year', values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(df, annot=True, linewidth=0.5, cmap='viridis')
```

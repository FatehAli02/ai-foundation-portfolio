### **Assessing and Cleaning Data**

**Data Assessing:** Understanding data deeply.

Types of Unclean data:
- Dirty data
    - Low quality data(Duplicated, missing values, corrupted, inaccurate data)

- Messy data
    - untidy data, has structural issues.
    
    These must be fulfilled:
    (each varible forms a column, each obs forms a row, each observational unit forms a table)
